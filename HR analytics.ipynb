## HR ANALYTICS (INDUSTRY LEVEL)

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
data = {
    "Employee":["Arun","Jai","Tripti","Simmy","Archita","Roshan"],
    "Department":["IT","HR","IT","Sales","HR","HR"],
    "Salary":[50000,45000,65000,52000,55000,39000],
    "Experience":[2,3,2,5,2,6],
    "Rating":[4.2,4.5,4.8,4.0,4.7,4.9]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
df.groupby("Employee")["Rating"].sum()
```
